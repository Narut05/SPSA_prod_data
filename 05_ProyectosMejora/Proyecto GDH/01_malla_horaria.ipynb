```python
df_unpivot2["hora_inicio"] = df_unpivot2["HORARIO"].str[0:5]
# df_unpivot2["hora_inicio"] = pd.to_datetime(
#     df_unpivot2["hora_inicio"],
#     format="%H:%M"
# ).dt.time

dic_turnos = {
    "CFR CAJERO 5X2 1ER TURNO-6": 1,
    "CFR CAJERO 5X2 1ER TURNO-7": 1,
    "CFR CAJERO 5X2 1ER TURNO-8": 1,
    "CFR CAJERO 5X2 1ER TURNO-9": 2,
    "CFR CAJERO 5X2 1ER TURNO-10": 2,
    "CFR CAJERO 5X2 1ER TURNO-11": 2,
    "CFR CAJERO 5X2 1ER TURNO-14": 3,
    "CFR CAJERO 5X2 1ER TURNO-17": 3,
    "CFR CAJERO 5X2 1ER TURNO-18": 3,
    "CFR CAJERO 5X2 2DO TURNO-7": 3,
    "CFR CAJERO 5X2 2DO TURNO-8": 3,
    "CFR CAJERO 5X2 2DO TURNO-9": 3,
    "CFR CAJERO 5X2 2DO TURNO-10": 3,
    "CFR CAJERO 5X2 2DO TURNO-11": 3,
    "CFR CAJERO 5X2 2DO TURNO-12": 3,
    "CFR CAJERO 5X2 2DO TURNO-13": 1,
    "CFR CAJERO 5X2 2DO TURNO-14": 1,
    "CFR CAJERO 5X2 2DO TURNO-15": 1,
    "CFR CAJERO 5X2 2DO TURNO-16": 2,
    "CFR CAJERO 5X2 2DO TURNO-17": 2,
    "CFR CAJERO 5X2 2DO TURNO-18": 2,
    "CFR CAJERO 5X2 2DO TURNO-19": 2,
    "CFR CAJERO 5X2 2DO TURNO-20": 2,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-6": 1,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-7": 1,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-8": 1,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-9": 2,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-10": 2,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-11": 2,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-12": 3,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-13": 3,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-14": 3,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-16": 3,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-17": 3,
    "CFR REPRES. DE SERVICIOS 5X2 1ER TURNO-18": 3,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-6": 3,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-7": 3,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-8": 3,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-9": 3,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-10": 3,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-11": 3,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-12": 3,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-13": 1,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-14": 1,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-15": 1,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-16": 2,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-17": 2,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-18": 2,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-19": 2,
    "CFR REPRES. DE SERVICIOS 5X2 2DO TURNO-20": 2
}

df_unpivot2["hora_inicio_2"] = df_unpivot2["hora_inicio"].astype(str).str[0:2].astype(int)
df_unpivot2["llave"] = df_unpivot2["puesto"].astype(str) + "-" + df_unpivot2["hora_inicio_2"].astype(str)
df_unpivot2["grupo_horario"] = df_unpivot2["llave"].map(dic_turnos)
df_unpivot3 = df_unpivot2.groupby(["TIENDA", "puesto", "dni", "grupo_horario"]).size().reset_index(name="cant")
# df_unpivot3["ratio_cant"] = df_unpivot3["cant"] / df_unpivot3.groupby(["TIENDA", "puesto", "dni"])["cant"].transform("sum")
df_unpivot3

df_pivot_final = df_unpivot3.pivot(
    index=["TIENDA", "puesto", "dni"],
    columns="grupo_horario",
    values="cant"
).fillna(0).reset_index()

df_pivot_final.columns

df_pivot_final["ratio1"] = df_pivot_final[1] / (df_pivot_final[1]+df_pivot_final[2]+df_pivot_final[3])
df_pivot_final["ratio2"] = df_pivot_final[2] / (df_pivot_final[1]+df_pivot_final[2]+df_pivot_final[3])
df_pivot_final["ratio3"] = df_pivot_final[3] / (df_pivot_final[1]+df_pivot_final[2]+df_pivot_final[3])

df_pivot_final["ratio_dominante"] = (
    df_pivot_final[["ratio1","ratio2","ratio3"]]
    .max(axis=1)
)

def clasificar(x):
    if x >= 0.90:
        return "1. Muy estable"
    elif x >= 0.80:
        return "2. Estable"
    elif x >= 0.50:
        return "3. Moderadamente variable"
    else:
        return "4. Muy variable"

df_pivot_final["clasificacion"] = (
    df_pivot_final["ratio_dominante"]
    .apply(clasificar)
)
```
