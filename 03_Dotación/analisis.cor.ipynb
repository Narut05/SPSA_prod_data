```python
import pandas as pd
```

```python
df1 = pd.read_excel(r"data - guardia civil.xlsx")
df2= pd.read_excel(r"data - jockey.xlsx")
df3 =  pd.read_excel(r"data - monterrico.xlsx")
df4 = pd.read_excel(r"data - san jorge.xlsx")
df5 = pd.read_excel(r"data - universitaria.xlsx")
```

```python
def analisis_corr (df):
    df1 = df.dropna()

    corr = (
        df1.groupby("Semana")
        .apply(
            lambda x: pd.Series({
                "Horario_vs_HorarioKr_DiaMes": x["Horario"].corr(x["HorarioKr_DiaMes"], method="pearson"),
                "Horario_vs_HorarioKr_Dia": x["Horario"].corr(x["HorarioKr_Dia"], method="pearson")
            })
        )
        .reset_index()
    )

    return corr
```

```python
analisis_corr (df1) #guardia civil
```

```python
def analisis_error(df):
    df["Error_horas"] = (
        df["HorarioKr_Dia"] - df["Horario"]
    ).abs()

    resumen = (
        df.groupby("Semana")
        .agg(
            ErrorPromedio=("Error_horas", "mean"),
            ErrorMediano=("Error_horas", "median"),
            ErrorMax=("Error_horas", "max")
        )
    )
    return resumen
```
